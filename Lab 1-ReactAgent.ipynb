```python
import math

# Minimal ReAct Agent

# -------------------------
# TOOL
# -------------------------

def unit_converter(value, from_unit, to_unit):

    conversions = {

        ("km", "m"): 1000,
        ("m", "km"): 0.001,

        ("kg", "g"): 1000,
        ("g", "kg"): 0.001,

    }

    key = (from_unit, to_unit)

    if key in conversions:

        return value * conversions[key]

    else:

        return "Conversion not supported"


# -------------------------
# REACT AGENT
# -------------------------

def react_agent(user_query):

    print("\n" + "✨" * 2 + " USER QUERY " + "✨" * 2 + "-" * 20)
    print(f"➜ {user_query}")
    print("-" * 40)

    # Step 1: Reasoning
    print("\n💡 **Thought:** I should use the unit converter tool to perform the conversion.")

    # Split sentence into words
    words = user_query.lower().split()

    value = None
    from_unit = None
    to_unit = None

    # Find number
    for word in words:

        if word.isdigit():

            value = int(word)
            break # Assume only one number for now

    # Find units
    units = ["km", "m", "kg", "g"]
    found_units = []

    for word in words:

        if word in units:

            found_units.append(word)

    if len(found_units) == 2:
        from_unit = found_units[0]
        to_unit = found_units[1]
    else:
        # Handle cases where units are not found or not two are provided
        print("⚠️ Could not identify both 'from' and 'to' units.")
        return # Exit if units are ambiguous

    # Step 2: Action
    print(f"\n⚙️ **Action:** Calling `unit_converter({value}, '{from_unit}', '{to_unit}')`")

    # Tool Call
    observation = unit_converter(value, from_unit, to_unit)

    # Step 3: Observation
    print(f"\n👀 **Observation:** The tool returned: {observation}")

    # Step 4: Final Answer
    print("\n" + "✅" * 2 + " FINAL ANSWER " + "✅" * 2 + "-" * 16)
    print(f"   {value} {from_unit} = {observation} {to_unit}")
    print("=" * 40)


# -------------------------
# RUN AGENT
# -------------------------

query = "Convert 5 km m"

react_agent(query)
```
